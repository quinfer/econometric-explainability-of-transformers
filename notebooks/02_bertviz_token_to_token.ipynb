# Visualising Token-to-Token Attention with BertViz

BertViz renders interactive diagrams that quantify how each token attends to others. This notebook frames the visual as a **conditional dependence matrix**, mirroring covariance heatmaps and similarity matrices used in econometrics.


## Setup

````python
from bertviz import head_view, model_view
from transformers import AutoTokenizer, AutoModel
import torch

MODEL_NAME = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModel.from_pretrained(MODEL_NAME, output_attentions=True)
````

## Choose an econometric narrative

The more structured your sentence, the easier it is to see which clauses attend to which. Try statements like:

> "Forward-looking Taylor rules anchor inflation expectations when credibility is high."


````python
text = "Forward-looking Taylor rules anchor inflation expectations when credibility is high."
inputs = tokenizer.encode_plus(text, return_tensors="pt")
with torch.no_grad():
    outputs = model(**inputs)
    attention = outputs.attentions

input_ids = inputs["input_ids"][0]
tokens = tokenizer.convert_ids_to_tokens(input_ids)
````

````python
head_view(attention, tokens)
````

### Interpreting the head view

- Each row is analogous to a regression where the dependent variable is a token embedding and the regressors are other tokens.
- Heads that strongly link "credibility" ↔ "high" behave like kernels concentrating mass on adjoining clauses.
- Compare across heads to identify complementarity, similar to examining multiple instruments or moments.


````python
model_view(attention, tokens)
````

## Econometric takeaways

- BertViz matrices behave like **weighting kernels**; you can read off which subclauses drive the conditional expectation at each position.
- Layer comparisons resemble checking robustness across specifications: if a dependency only shows up late, probe whether it is essential (akin to testing over-identifying restrictions).
- Use the diagrams in lectures as a visual analogue to covariance/correlation heatmaps students already know.
